**🧾 Column Descriptions**

| **Column Name** | **Description**                                                        |
| --------------- | ---------------------------------------------------------------------- |
| **BrandName**   | The brand or manufacturer of the product.                              |
| **Details**     | Descriptive details about the product (style, material, design, etc.). |

```python
## Importing important libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## Loading the  dataset
df = pd.read_csv("FashionDataset.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
## Renaming each column properly
df = df.rename(columns={'Unnamed: 0':"Sales_id", 'BrandName':"Brand", 'Deatils':"details",
       'Sizes':"Sizes", 'MRP':"MRP", 'SellPrice':"Selling price",
       'Discount':"Discount in %", 'Category':"Category"})
```

```python
df.columns
```

```python
## Ordering columns in the right order
right_Order = ["Sales_id","details","Category","Sizes","Brand","MRP","Selling price","Discount in %"]
df = df[right_Order]
```

```python
df
```

```python
## extracting MRP from MRP column
df["MRP"] = df["MRP"].str.replace("Rs\n","")
```

```python
## extracting discount fro m Discount in % column
df["Discount in %"] = df["Discount in %"].str.replace("% off","")
```

```python
df["Category"].value_counts()
```

```python
## Removing -Women String from category column
df["Category"] = df["Category"].str.replace("-Women","")
```

```python
df.info()
```

Here, It is showing there are no null values because the null values are present as string values "Nan". we need to replace them with Numpy NaN

```python
df = df.replace("Nan", np.nan)
```

```python
df
```

```python
## converting MRP, Selling price & discount columns as Float

df[["MRP","Selling price","Discount in %"]] = df[["MRP","Selling price","Discount in %"]].\
astype("float")
```

```python
df.info()
```

```python
## Checking for duplicate records
df.duplicated().sum() ## There are 1331 duplicate records
```

```python
## Dropping duplicate records
df.drop_duplicates(inplace=True)
```

```python
## Checking for null values

df.isna().sum()
```

```python
df["details"].value_counts()
```

```python
df["Sizes"].value_counts()
```

```python
df["Brand"].value_counts()
```

```python
sns.boxplot(data=df,x="Selling price")
```

```python
sns.boxplot(data=df,x="Discount in %")
```

Here, 
details column can be dropped because it has no much significance.
Sizes column can be dropped becuase there is no consistency in Size categories.
In Brand column null values can be replaced by Mode.
In Discount column null values can be replaced by Mean as it haas no outliers
the MRP & Selling price null values can be replaced by median since they have high number of outliers

```python
## Dropping details & sizes column

df = df.drop(columns=["details","Sizes"])
```

```python
## replacing Brand column null values with mode
df["Brand"].mode()
```

```python
df["Brand"] = df["Brand"].fillna("vastranand")
```

```python
## replacing Discount column null values with mean
df["Discount in %"] = df["Discount in %"].fillna(df["Discount in %"].mean())
```

```python
## replacing MRP column null values with Median
df["MRP"] = df["MRP"].fillna(df["MRP"].mean())
```

```python
## replacing Selling price column null values with Median
df["Selling price"] = df["Selling price"].fillna(df["Selling price"].mean())
```

```python
df.isna().sum() ## Now, there are no more null values
```

## Category Analysis

```python
## Top categories by number of products sold
df["Category"].value_counts()
```

Westernwear, Indianwear are the categories with highest number of products sold,9962 & 9858 respectively covering 2/3 rd of total products sold



```python
## Top categories with highest total sales
Category_sales = df.groupby("Category")["Selling price"].sum()
Category_sales = pd.DataFrame(Category_sales)
Category_sales["Selling price"] = round(Category_sales["Selling price"],2)
Category_sales.sort_values(by="Selling price",ascending=False)
```

Watches has the sales of 1.5 Crore with just 1725 products

```python
## Categories with highest & Lowest Avg discount
Category_discount = df.pivot_table(index="Category",values="Discount in %",aggfunc="mean")
Category_discount.sort_values(by="Discount in %",ascending=False)
```

Indianwear has the highest Avg discount of 49.6%
Fragrance has the highest Avg discount of 36%

```python
## Category with costliest product
Category_max_price = df.pivot_table(index="Category",values="Selling price",aggfunc="max")
Category_max_price.sort_values(by="Selling price",ascending=False)
```

Watches has the costliest product with the price of 1.4 Lakh

## Brand Analysis

```python
## Brands with highest sales
Top_selling_brand = df.pivot_table(index="Brand",values="Selling price",aggfunc="sum")
Top_selling_brand["Selling price"] = round(Top_selling_brand["Selling price"],2)
Top_selling_brand.sort_values(by="Selling price",ascending=False)
```

vastranand - Indian wear brand has highest sales - 32 Lakhs

```python
## Brands with costliest products
Costly_brand = df.pivot_table(index="Brand",values="Selling price",aggfunc="max")
Costly_brand.sort_values(by="Selling price",ascending=False).head(3)
```

versace, tissot, michael kors are tthe Watch brand with costliest items

```python
## Brands with highest Avg discount
Brand_discount = df.pivot_table(index="Brand",values="Discount in %",aggfunc="mean")
Brand_discount.sort_values(by="Discount in %",ascending=False).head()
```

gas - an indian wear brand has highest Avg Discount

We need to export the data of each Category into seperate files

```python
## Creating dataframes for each category

Westernwear_data = df[df["Category"]=="Westernwear"]
Indianwear_data = df[df["Category"]=="Indianwear"]
Lingerie_Nightwear_data = df[df["Category"]=="Lingerie&Nightwear"]
Footwear_data = df[df["Category"]=="Footwear"]
Watches_data = df[df["Category"]=="Watches"]
Fragrance_data = df[df["Category"]=="Fragrance"]
Jewellery_data = df[df["Category"]=="Jewellery"]
```

```python
## Exporting Data of each category

Westernwear_data.to_csv("Westernwear_data.csv")
Indianwear_data.to_csv("Indianwear_data.csv")
Lingerie_Nightwear_data.to_csv("Lingerie_Nightwear_data.csv")
Footwear_data.to_csv("Footwear_data.csv")
Watches_data.to_csv("Watches_data.csv")
Fragrance_data.to_csv("Fragrance_data.csv")
Jewellery_data.to_csv("Jewellery_data.csv")
```
